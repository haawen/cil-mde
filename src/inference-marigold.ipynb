```python
import os
import glob
import random
import numpy as np
import cv2
from PIL import Image
import torch
import diffusers  # Diffusers' Marigold pipeline
```

```python
train_dir = "/kaggle/input/ethz-cil-monocular-depth-estimation-2025/train/train"
test_dir  = "/kaggle/input/ethz-cil-monocular-depth-estimation-2025/test/test"
output_width, output_height = 560, 426
output_npy_dir = "/kaggle/working/marigold_pipe_preds_npy"
output_img_dir = "/kaggle/working/marigold_pipe_preds_img"
os.makedirs(output_npy_dir, exist_ok=True)
os.makedirs(output_img_dir, exist_ok=True)
```

```python
pipe = diffusers.MarigoldDepthPipeline.from_pretrained("prs-eth/marigold-depth-v1-0", variant="fp16",torch_dtype=torch.float16)
pipe = pipe.to("cuda")
```

```python
"""
all_rgb_paths = glob.glob(os.path.join(train_dir, "*_rgb.png"))
base_names = [os.path.basename(path).replace("_rgb.png", "") for path in all_rgb_paths]

# Randomly sample a subset (e.g., up to 1000 samples for calibration)
num_samples = 400
sample_names = random.sample(base_names, min(num_samples, len(base_names)))

all_preds = []  # To store flattened raw predictions from Marigold
all_gts = []    # To store flattened ground truth depths

print("Beginning calibration on training samples...")
for name in sample_names:
    rgb_path = os.path.join(train_dir, f"{name}_rgb.png")
    depth_path = os.path.join(train_dir, f"{name}_depth.npy")
    
    # Load image with PIL (the pipeline expects a PIL RGB image)
    try:
        image = Image.open(rgb_path).convert("RGB")
    except Exception as e:
        print(f"Error loading image {name}: {e}")
        continue

    # Run inference using the Marigold pipeline
    try:
        result = pipe(image)
    except Exception as e:
        print(f"Error processing image {name} through pipeline: {e}")
        continue

    # Get the raw depth prediction; according to the HF docs, the output has a property "prediction"
    raw_depth = result.prediction  
    raw_depth_np = np.array(raw_depth)
    raw_depth_np = np.squeeze(raw_depth_np)
    
    # Check that the prediction is valid (should be a 2D array)
    if raw_depth_np.size == 0 or raw_depth_np.ndim != 2:
        print(f"Warning: Skipping training sample {name} due to invalid prediction shape: {raw_depth_np.shape}")
        continue

    # Resize raw depth to target dimensions (560×426)
    try:
        raw_depth_resized = cv2.resize(raw_depth_np, (output_width, output_height))
    except Exception as e:
        print(f"Error resizing training image {name}: {e}")
        continue

    # Load the corresponding ground truth depth (assumed to be in meters)
    try:
        gt_depth = np.load(depth_path)
    except Exception as e:
        print(f"Error loading ground truth for {name}: {e}")
        continue

    # Flatten and store for calibration
    all_preds.append(raw_depth_resized.flatten())
    all_gts.append(gt_depth.flatten())

if len(all_preds) == 0:
    raise ValueError("No valid training predictions were collected for calibration.")

# Concatenate all flattened arrays
preds_flat = np.concatenate(all_preds)
gts_flat = np.concatenate(all_gts)

# Perform a least-squares linear regression to compute calibration parameters:
# Find a and b such that: gt_depth ≈ a * raw_depth + b
A = np.vstack([preds_flat, np.ones_like(preds_flat)]).T
a, b = np.linalg.lstsq(A, gts_flat, rcond=None)[0]
print(f"Calibration complete: scale (a) = {a}, shift (b) = {b}")

#scale (a) = 2.967557191848755, shift (b) = 1.6479114294052124
"""
a = 2.967557191848755
b = 1.6479114294052124
```

```python
with open("/kaggle/input/ethz-cil-monocular-depth-estimation-2025/test_list.txt", "r") as f:
    test_list = [line.strip().split()[0] for line in f]

print("Running inference on test images...")
for filename in test_list:
    img_path = os.path.join(test_dir, filename)
    try:
        image = Image.open(img_path).convert("RGB")
    except Exception as e:
        print(f"Error loading test image {filename}: {e}")
        continue

    try:
        result = pipe(image)
    except Exception as e:
        print(f"Error processing test image {filename} through pipeline: {e}")
        continue

    raw_depth = np.array(result.prediction)
    raw_depth_np = np.squeeze(raw_depth)
    if raw_depth_np.size == 0 or raw_depth_np.ndim != 2:
        print(f"Warning: Skipping test image {filename} due to invalid prediction shape: {raw_depth.shape}")
        continue

    try:
        depth_resized = cv2.resize(raw_depth_np, (output_width, output_height))
    except Exception as e:
        print(f"Error resizing test image {filename}: {e}")
        continue

    # Apply calibration: metric_depth = a * raw_depth + b
    depth_calibrated = a * depth_resized + b
    # Clip any negative values (set minimum to 0.1 m)
    depth_calibrated = np.maximum(depth_calibrated, 0.1)

    # Save the calibrated depth map as a .npy file.
    base = filename.replace("_rgb.png", "_depth.npy")
    save_path_npy = os.path.join(output_npy_dir, base)
    np.save(save_path_npy, depth_calibrated)
    
    # Create a visualization image:
    depth_vis = cv2.normalize(depth_calibrated, None, 0, 255, cv2.NORM_MINMAX)
    depth_vis = np.uint8(depth_vis)
    depth_vis = cv2.applyColorMap(depth_vis, cv2.COLORMAP_JET)  # use the 'Spectral' colormap
    save_path_img = os.path.join(output_img_dir, base.replace(".npy", ".png"))
    cv2.imwrite(save_path_img, depth_vis)
    
    print(f"Processed {filename}: saved calibrated depth and visualization.")
```
